```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully.")
```

```python
print("First 10 records:")
display(df.head(10))

print("Dataset shape: ", df.shape)
print("Column names:",df.columns.tolist())
```

```python
print("Dataset information: ")
df.info()

print("\\nMissing values in each column: ")
display(df.isnull().sum())

print("\\nStatistical Summary: ")
display(df.describe())
```

```python
df.insert(0,"student_ID", range(1,len(df) + 1))
subjects = ["math score","reading score","writing score"]
df["Total_Marks"] = df[subjects].sum(axis=1)
df["Average_Marks"] = df[subjects].mean(axis=1)
display(df[["Student_ID"] + subjects + ["Total_Marks", "Average_Marks"]].head(10))
```

```python
marks = df[subjects].to_numpy()
print("Maximum score:",np.max(marks))
print("Minimum score:",np.min(marks))
print("Mean score: ",round(np.mean(marks),2))
print("Median score:", round(np.median(marks),2))
print("Standard devition: ",round(np.std(marks),2))

for subject in subjects:
  values = df[subject].to_numpy()
  print(
      subject,
      "| Max:",round(np.max(values),2),
      "| Min:",round(np.min(values),2),
      "| Mean:",round(np.mean(values),2),
      "| Median:",round(np.median(values),2),
  )
```

```python
top_10 = df.nlargest(10,"Average_Marks")
display(top_10[["Student_ID","Average_Marks"]])
```

```python
below_40 = df[(df[subjects] < 40).any(axis=1)]

print("Number of student scorring below 40 in  at least one subject:", len(below_40))
display(below_40[["Student_ID","gender"] + [subject]])
```

```python
def categorize_performance(avg):
  if avg >= 80:
    return "Excellent"
  elif avg >= 70:
    return "Good"
  elif avg >= 50:
    return "Average"
  else:
    return "Needs Improvement"
  df["Performance"] = df["Average_Marks"].apply(categorize_performance)
  display(df[["Student_ID","Average_Marks","Performance"]].head(15))
  print("Number of students in each category:")
  display(df["Performance"].value_counts())
```
